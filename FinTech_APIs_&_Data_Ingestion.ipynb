## Problem 1: Make a Basic API Call

**Task:** Complete the function to fetch all users from the JSONPlaceholder API and return the JSON response.

**API Endpoint:** `https://jsonplaceholder.typicode.com/users`

**Requirements:**
- Make a GET request to the URL
- Return the JSON data if status code is 200
- Return None if request fails

```python
def fetch_all_users():
    """
    Fetch all users from JSONPlaceholder API.

    Returns:
        list: List of user dictionaries if successful, None otherwise
    """
    url = "https://jsonplaceholder.typicode.com/users"

    # YOUR CODE HERE
    # 1. Make a GET request to the URL using requests.get()
    response = requests.get(url)
    # 2. Check if response.status_code == 200
    if response.status_code == 200:
      return response.json()
    # 3. If yes, return response.json()
    else:
      return None
    # 4. If no, return None



# Test your function
users = fetch_all_users()
assert users is not None, "Failed to fetch users"
assert len(users) == 10, f"Expected 10 users, got {len(users)}"
assert 'name' in users[0], "User data missing 'name' field"
print("✅ Problem 1 passed!")
print(f"Fetched {len(users)} users")
```

## Problem 2: Navigate Nested JSON

**Task:** Extract specific information from nested JSON data.

- User's email
- User's city (nested in address)
- User's company name (nested in company)

**Requirements:**

```python
def extract_user_info(user):
    """
    Extract key information from a user object.

    Args:
        user (dict): User dictionary with nested data

    Returns:
        dict: Dictionary with 'name', 'email', 'city', 'company_name'
    """

    # YOUR CODE HERE
    # Example user structure:
    # {
    #   "name": "Leanne Graham",
    #   "email": "Sincere@april.biz",
    #   "address": {
    #     "city": "Gwenborough"
    #   },
    #   "company": {
    #     "name": "Romaguera-Crona"
    #   }
    # }
    #
    # 1. Extract user['name']
    # 2. Extract user['email']
    # 3. Extract user['address']['city']
    # 4. Extract user['company']['name']
    # 5. Return a dictionary with keys: 'name', 'email', 'city', 'company_name'

    user = {
        'name' : user['name'],
        'email' : user['email'],
        'city' : user['address']['city'],
        'company_name' : user['company']['name']

    }

    return user



# Test your function
users = fetch_all_users()
info = extract_user_info(users[0])
assert 'name' in info, "Missing 'name' key"
assert 'email' in info, "Missing 'email' key"
assert 'city' in info, "Missing 'city' key"
assert 'company_name' in info, "Missing 'company_name' key"
assert info['name'] == 'Leanne Graham', f"Expected 'Leanne Graham', got {info['name']}"
print("✅ Problem 2 passed!")
print(f"Extracted info: {info}")
```

## Problem 9: Data Aggregation

**Task:** Fetch posts and create a summary DataFrame showing post count per user.

**API Endpoint:** `https://jsonplaceholder.typicode.com/posts`

**Requirements:**
- Fetch all posts
- Convert to DataFrame
- Group by 'userId' and count posts
- Return DataFrame with columns: 'user_id' and 'post_count'
- Sort by post_count in descending order

```python
def get_post_summary():
    """
    Get summary of post count per user.

    Returns:
        DataFrame: DataFrame with 'user_id' and 'post_count' columns
    """
    url = "https://jsonplaceholder.typicode.com/posts"

    # YOUR CODE HERE
    # 1. Fetch all posts from the API
    posts = requests.get(url).json()
    # 2. Convert to DataFrame: df = pd.DataFrame(posts)
    df = pd.DataFrame(posts)
    # 3. Group by 'userId' and count: df.groupby('userId').size()
    grouped = df.groupby('userId').size()
    # 4. Convert result to DataFrame with reset_index()
    results = grouped.reset_index()
    # 5. Rename columns to 'user_id' and 'post_count'
    results.columns = ['user_id','post_count']
    # 6. Sort by 'post_count' in descending order
    results = results.sort_values(by='post_count',ascending=False)
    # 7. Return the DataFrame
    return results

    #pass  # Remove this line when you write your code


# Test your function
summary = get_post_summary()
assert isinstance(summary, pd.DataFrame), "Should return a DataFrame"
assert list(summary.columns) == ['user_id', 'post_count'], "Columns should be 'user_id' and 'post_count'"
assert len(summary) == 10, f"Expected 10 users, got {len(summary)}"
assert summary['post_count'].iloc[0] >= summary['post_count'].iloc[-1], "Should be sorted in descending order"
print("✅ Problem 9 passed!")
print(f"\nPost count summary:")
print(summary)
```

## Problem 10: Real-World Pipeline

**Task:** Build a complete data pipeline that combines users and their posts.

**Requirements:**
- Fetch all users and all posts
- Convert both to DataFrames
- Merge them on user ID
- Select columns: 'user_name', 'user_email', 'post_title', 'post_body'
- Return the first 20 rows

This is a comprehensive problem that tests your understanding of the entire workflow!

```python
def create_user_posts_pipeline():
    """
    Create a complete pipeline merging users and posts.

    Returns:
        DataFrame: Merged DataFrame with user and post information
    """

    # YOUR CODE HERE
    # Step 1: Fetch users
    # - Use requests.get() to fetch from 'https://jsonplaceholder.typicode.com/users'
    url1 = 'https://jsonplaceholder.typicode.com/users'
    users = requests.get(url1).json()

    # - Convert to DataFrame
    df_users  = pd.DataFrame(users)[['id','name','email']]
    # - Select columns: ['id', 'name', 'email']
    df_users = df_users.rename(columns={'id':'user_id','name':'user_name','email':'user_email'})
    # - Rename to: ['user_id', 'user_name', 'user_email']

    # Step 2: Fetch posts
    # - Use requests.get() to fetch from 'https://jsonplaceholder.typicode.com/posts'
    url2 = 'https://jsonplaceholder.typicode.com/posts'
    posts = requests.get(url2).json()
    # - Convert to DataFrame
    df_post = pd.DataFrame(posts)[['userId','title','body']]
    # - Select columns: ['userId', 'title', 'body']
    df_post = df_post.rename(columns={'userId':'user_id','title':'post_title','body':'post_body'})
    # - Rename to: ['user_id', 'post_title', 'post_body']

    # Step 3: Merge DataFrames
    # - Use pd.merge(df_users, df_posts, on='user_id')
    df_merged = pd.merge(df_users, df_post, on='user_id')

    # Step 4: Select final columns
    # - Select: ['user_name', 'user_email', 'post_title', 'post_body']
    df_merged = df_merged[['user_name', 'user_email', 'post_title', 'post_body']]

    # Step 5: Return first 20 rows
    return df_merged.head(20)
    # - Use .head(20)




# Test your function
pipeline_df = create_user_posts_pipeline()
assert isinstance(pipeline_df, pd.DataFrame), "Should return a DataFrame"
expected_columns = ['user_name', 'user_email', 'post_title', 'post_body']
assert list(pipeline_df.columns) == expected_columns, f"Expected columns {expected_columns}"
assert len(pipeline_df) == 20, f"Expected 20 rows, got {len(pipeline_df)}"
assert pipeline_df['user_name'].notna().all(), "user_name should not have null values"
print("✅ Problem 10 passed!")
print(f"\nFirst 5 rows of merged data:")
print(pipeline_df.head())
```
