```python
# --- 1. LISTS (The Ledger) ---
# Ordered, changeable, and allows duplicates. Wrapped in square brackets [].
daily_payouts = [5000.00, 1200.50, 5000.00, 890.00, 45000.00]

print(f"First payout of the day: ₹{daily_payouts[0]}") # Python starts counting at 0!


# --- 2. SETS (The Deduplicator) ---
# Unordered and strictly UNIQUE. Wrapped in curly braces {}.
# Imagine we have a messy list of Merchant IDs that hit our gateway, and we need to know exactly who was active today without counting them twice.
messy_merchant_pings = ["M_101", "M_102", "M_101", "M_101", "M_103", "M_102"]

unique_active_merchants = set(messy_merchant_pings)
print(f"Unique merchants active today: {unique_active_merchants}")


# --- 3. DICTIONARIES (The Payload) ---
# Key-Value pairs. Wrapped in curly braces {}, but structured with a colon.
# When Visa or Razorpay sends you a transaction, it looks EXACTLY like this.
transaction_payload = {
    "txn_id": "pay_xyz123",
    "user_email": "aman@email.com",
    "amount": 15000.00,
    "currency": "INR",
    "is_international": False
}

# Pulling a specific datapoint out of the payload
print(f"Transaction ID {transaction_payload['txn_id']} was processed for ₹{transaction_payload['amount']}.")
```

## Module 3: Functions (Building Reusable Machines)

A **Function** is the ultimate tool for automation. Instead of writing the same calculation 100 times, you build a "machine" once, give it a name, and then just push raw data through it whenever you need to.

**The Business Need:** We need to calculate the Net Settlement Amount for our merchants (Transaction Amount minus the 1.5% gateway fee).

```python
# 1. BUILDING THE MACHINE (Define the function using 'def')
def calculate_net_settlement(gross_amount):
    """
    This function takes a gross transaction amount,
    deducts a 1.5% fee, and returns the net payout.
    """
    fee_rate = 0.015
    fee_amount = gross_amount * fee_rate
    net_payout = gross_amount - fee_amount

    # 'return' is what the machine spits out at the end
    return round(net_payout, 2)


# 2. USING THE MACHINE (Calling the function)
# We can now push any number through our new machine instantly.
zomato_gross = 1000.00
flipkart_gross = 45000.00

zomato_net = calculate_net_settlement(zomato_gross)
flipkart_net = calculate_net_settlement(flipkart_gross)

print("--- DAILY SETTLEMENT REPORT ---")
print(f"Zomato Gross: ₹{zomato_gross} -> Net Payout: ₹{zomato_net}")
print(f"Flipkart Gross: ₹{flipkart_gross} -> Net Payout: ₹{flipkart_net}")
```

## Module 4: Visualization (The Executive Dashboard)

If you hand a CEO a list of 50,000 numbers, they will ask you to leave the room. Data is only valuable when it tells a story.

In Python, we don't manually drag and drop charts like in Excel. We use a **Library**—a pre-built toolkit of code written by other developers.

**Matplotlib** is the industry standard for Python data visualization. It is our digital paintbrush. We give it two lists (our X-axis and our Y-axis), tell it what kind of chart we want, and it instantly paints it.

```python
# --- 1. IMPORTING THE TOOLKIT ---
# We bring in matplotlib. 'plt' is just the universal industry nickname for it so we don't have to type the full name every time.
import matplotlib.pyplot as plt

# --- 2. THE RAW DATA (Our Lists) ---
# Imagine this is our daily processing volume for the Diwali week.
days_of_week = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
daily_volume_inr = [120000, 135000, 150000, 290000, 310000, 180000, 140000]

# --- 3. SETTING UP THE CANVAS ---
# We tell Python how big the chart should be (Width, Height)
plt.figure(figsize=(10, 5))

# --- 4. PAINTING THE CHART ---
# We use a line plot. We pass our X list, then our Y list.
# marker='o' puts a clear dot on every single data point.
plt.plot(days_of_week, daily_volume_inr, color="navy", marker="o", linewidth=2, label="Processing Volume")

# Adding the Business Context
plt.title("Diwali Week: Daily Processing Volume", fontsize=14, fontweight="bold")
plt.xlabel("Day of the Week")
plt.ylabel("Volume (₹)")

# The Power Move: Adding a visual risk threshold
# Let's say our gateway crashes if we cross ₹250,000 in a day. We draw a red line at that exact limit.
plt.axhline(y=250000, color="red", linestyle="--", label="System Capacity Limit")

# --- 5. FINISHING TOUCHES ---
plt.legend()       # Shows the key/legend
plt.grid(alpha=0.3) # Adds a faint background grid so executives can trace the numbers easily

# 'show()' is the command that actually renders the image to the screen.
plt.show()
```
